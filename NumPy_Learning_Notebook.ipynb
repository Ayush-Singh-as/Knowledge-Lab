# 1. NumPy Basics

An one-dimensional NumPy Array

```python
a = [1, 2, 3]
np.array(a)
```

A 2d NumPy Array

```python
lst = [[1, 2, 3],
       [4, 5, 6]]
array2d = np.array(lst)
array2d
```

Lets check the data type of the array -

```python
array2d.dtype
```

How to define an array of specific data type like int32?

```python
int32Array = np.array(lst, dtype=np.int32)
int32Array
```

We can also downcast or recast array's data type using the astype method like -

```python
int32array = array2d.astype(np.int32)
int32array
```

To return the number of elements in an array, we can use the size attribute, as shown below:



```python
array2d.size
```

And the number of dimensions of our array (Intuitively, you may think of dimensions as the rank of a tensor) can be obtained via the ndim attribute:

```python
array2d.ndim
```

To return the number of elements along each array dimension (in 2d array it will be number of rows and columns) we can use the shape attribute as shown below:

```python
array2d.shape
```
