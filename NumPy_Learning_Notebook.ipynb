# 3. NumPy Array Math and Universal Functions

In the previous sections, you learned how to create NumPy arrays and how to access different elements in an array. It is about time that we introduce one of the core features of NumPy that makes working with ndarray so efficient and convenient: vectorization. While we typically use for-loops if we want to perform arithmetic operations on sequence-like objects, NumPy provides vectorized wrappers for performing element-wise operations implicitly via so-called ufuncs – “ufuncs” is short for universal functions.

Check out the [official documentation](https://docs.scipy.org/doc/numpy/reference/ufuncs.html#available-ufuncs) for a complete list of ufuncs available in NumPy.

The ufuncs for basic arithmetic operations are add, subtract, divide, multiply, power, and exp (exponential). However, NumPy uses operator overloading so that we can use mathematical operators (+, -, /, *, and **) directly:

```python
lst = [[1, 2, 3],
        [4, 5, 6]]
ary = np.array(lst)
np.add(ary, 1)
```

```python
ary + 1
```

In addition, NumPy implements several useful unary ufuncs, such as log (natural logarithm), log10 (base-10 logarithm), and sqrt (square root):

```python
np.log(ary)
```

Often, we want to compute the sum or product of array element along a given axis. For this purpose, we can use a ufunc’s reduce operation. By default, reduce applies an operation along the first axis (axis=0):

```python
np.add.reduce(ary)
```

```python
np.add.reduce(ary, axis=1)
```

NumPy also provides specific functions such as sum and product:

```python
ary.sum(axis=0)
```

Note: Both product and sum function compute the product and sum of the entire array if we do not define an axis:

```python
ary.sum()
```

Other useful unary ufuncs are:

* np.mean (computes arithmetic mean or average)
* np.std (computes the standard deviation)
* np.var (computes variance)
* np.sort (sorts an array)
* np.argsort (returns indices that would sort an array)
* np.min (returns the minimum value of an array)
* np.max (returns the maximum value of an array)
* np.argmin (returns the index of the minimum value)
* np.argmax (returns the index of the maximum value)
* np.array_equal (checks if two arrays have the same shape and elements)

# 4. NumPy Broadcasting

Broadcasting allows us to perform vectorized operations between two arrays even if their dimensions do not match by creating implicit multidimensional grids.

In the previous section, we learned about ufuncs by performing element-wise addition between a scalar and a multidimensional array, which is one example of broadcasting:

```python
ary = np.array([1, 2, 3])
ary + 1
```

We can also perform element-wise operations between arrays of equal dimensions:

```python
ary + np.array([1, 1, 1])
```

Unlike standard linear algebra, NumPy allows addition of arrays with different shapes.
In this example, a one-dimensional array is added to a two-dimensional array by implicitly expanding it into a multidimensional grid:

```python
ary + np.array([[4, 5, 6],
                [7, 8, 9]])
```

# 5. NumPy Advanced Indexing -  Memory Views and Copies

## Advanced Indexing - Memory Views and Copies

In earlier sections, we used basic indexing and slicing. These operations create views of NumPy arrays, not copies, meaning they share the same memory. This is useful because it saves memory by avoiding unnecessary data duplication. To understand this, consider a simple example where we select the first row of an array, store it in a variable, and then modify it:

```python
ary = np.array([[1, 2, 3],
                [4, 5, 6]])
firstRow = ary[0]
firstRow
```

```python
firstRow += 10
firstRow
```

Note that the original array was modified as well:

```python
ary
```

The same concept applies to slicing operations:

```python
secondRow = ary[1:3]
secondRow += 10
ary
```

However, in certain scenarios we want force a copy of an array; we can do this via the copy method as shown below:

```python
firstRow = ary[0].copy()
firstRow += 10
firstRow
```

```python
ary
```

## Fancy Indexing

Besides basic single-integer indexing and slicing, NumPy also provides advanced indexing methods known as *fancy indexing*. Fancy indexing uses lists or tuples of non-contiguous integer indices to select specific elements from an array. Because these elements are not stored contiguously in memory, fancy indexing cannot return a view and therefore always produces a copy of the array. The following examples demonstrate this behavior:

```python
ary = np.array([[1, 2, 3],
              [4, 5, 6]])
ary[:, [0, 2]]
```

## Boolean Masks for Indexing

Finally, we can also use Boolean masks for indexing – that is, arrays of True and False values. Consider the following example, where we return all values in the array that are greater than 4:

```python
ary = np.array([[1, 2, 3],
                [4, 5, 6]])
greaterMask = ary > 4
greaterMask
```

Using these masks, we can select elements given our desired criteria:

```python
ary[greaterMask]
```

We can also chain different selection criteria using the logical and operator & or the logical or operator | :

```python
ary[(ary > 4) | (ary%2 == 0)]
```

Note that indexing using Boolean arrays is also considered “fancy indexing” and thus returns a copy of the array.

# 6. Random Number Generators

The `numpy.random` module provides pseudo-random number generators (PRNGs) that allow users to draw samples from a wide range of probability distributions. Typically, a `Generator` instance is created using `default_rng`, and its methods are then used to generate samples from different distributions.

Let's generate array of random floats uniformly distributed over the range [0,1):

```python
rng = np.random.default_rng(67)
rng.random(3)
```

Let's generate an array of 10 numbers according to a unit Gaussian distribution:

```python
rng.standard_normal(10)
```

Let's generate an array of 5 integers uniformly over the range [0, 10):

```python
rng.integers(low=0, high=10, size=5)
```

NumPy’s random number generators are deterministic and can be reproduced by setting a seed. If no seed is provided, `default_rng` uses nondeterministic data from the operating system, producing different results each time. Seeds should be large positive integers, and using large, unique values is recommended to ensure statistical independence. A convenient way to generate such a seed is with `secrets.randbits`, for example to create a 128-bit integer.

```python
import secrets
secrets.randbits(128)
```

```python
rng = np.random.default_rng(secrets.randbits(128))
rng.random()
```

# 7. Reshaping NumPy Arrays

In practice, arrays often do not have the required shape for a computation. Although the size of a NumPy array is fixed, its shape can be changed without copying data. NumPy provides the `reshape` method to create a new view of the same array with a different shape. For example, a one-dimensional array can be reshaped into a two-dimensional array using `reshape`:

```python
array_1d = np.array([1, 2, 3, 4, 5, 6])
view_in_2d = array_1d.reshape(2, 3)
view_in_2d
```

The True value returned by `np.may_share_memory` shows that the `reshape` operation produces a view of the original array rather than a copy:

```python
np.may_share_memory(view_in_2d, array_1d)
```

When reshaping an array, the total number of elements must remain the same as in the original array. However, it is not necessary to specify the size of every axis, since NumPy can automatically infer the missing dimension when one axis is set to -1:

```python
array_1d.reshape(-1, 3)
```

We can also use `reshape` to flatten an array:

```python
array_2d = np.array([[1, 2, 3],
                     [4, 5, 6]])
array_2d.reshape(-1)
```

Other ways to flatten arrays include `flatten`, which creates a copy of the array, and `ravel`, which returns a memory view similar to `reshape`:

```python
array_2d.flatten()
```

```python
array_2d.ravel()
```

Sometimes it is necessary to merge multiple arrays, even though NumPy arrays have a fixed size and combining them always creates a new array. Because this operation can be computationally expensive, it should be used sparingly. NumPy provides the `concatenate` function to join two or more arrays, as shown in the following examples:

```python
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
np.concatenate((array1, array2))
```

```python
array1 = np.array([[1, 2, 3]])
array2 = np.array([[4, 5, 6]])
np.concatenate((array1, array2))
```

```python
array1 = np.array([[1, 2, 3]])
array2 = np.array([[4, 5, 6]])
np.concatenate((array1, array2), axis=1)
```
