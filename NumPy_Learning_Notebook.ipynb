# 2. NumPy Array Construction and Indexing

## Array Construction

Constructing an array containing ones or zeroes only:

```python
np.ones((3, 3))
```

```python
np.zeros((3, 3))
```

We can use these functions to create arrays with arbitrary value:

```python
np.zeros((3, 3)) + 5
```

We can also create an empty array which can act as a placeholder array:

```python
np.empty((3, 3))
```

Note: An empty array will contain whatever random values were already sitting in the memory.

NumPy also comes with functions to create identity matrices and diagonal matrices as ndarrays:

```python
np.eye(3)
```

```python
np.diag((1, 2, 3))
```

Lastly, I want to mention two very useful functions for creating sequences of numbers within a specified range, namely, arange and linspace:

```python
np.arange(1, 10)
```

If we only provide a single function argument, the range object treats this number as the endpoint of the interval and starts at 0:

```python
np.arange(10)
```

In NumPy's arange function, a third argument can be provided to define the step (the default step size is 1):*italicized text*

```python
np.arange(1, 2, 0.1)
```

PS: NumPy's arange function is similar to python's range objects.

The linspace function is especially useful if we want to create a particular number of evenly spaced values in a specified half-open interval:

```python
np.linspace(1, 10, num=10)
```

```python
np.linspace(1, 10, num=17)
```

```python
np.linspace(1, 10)
```

## Array Indexing

NumPy indexing and slicing works similar to Python lists, which we will demonstrate in the following code snippets:

```python
ary = np.array([1, 2, 3])
ary[1]
```

Indexing a NumPy array gives a NumPy number (like np.int64) instead of a normal Python number so it matches the array’s data type.

```python
ary[0:2] # Slicing
```

If we work with arrays that have more than one dimension or axis, we separate our indexing or slicing operations by commas as shown in the series of examples below:

```python
ary = np.array([[1, 2, 3],
                [4, 5, 6]])
ary[0, 2] # first row last column
```

```python
ary[1, 1] # second row second column
```

```python
ary[-1, -1] # last row last column
```

```python
ary[:, 0] # entire first column
```

```python
ary[:, :2] # first two columns
```

# 3. NumPy Array Math and Universal Functions

In the previous sections, you learned how to create NumPy arrays and how to access different elements in an array. It is about time that we introduce one of the core features of NumPy that makes working with ndarray so efficient and convenient: vectorization. While we typically use for-loops if we want to perform arithmetic operations on sequence-like objects, NumPy provides vectorized wrappers for performing element-wise operations implicitly via so-called ufuncs – “ufuncs” is short for universal functions.

Check out the [official documentation](https://docs.scipy.org/doc/numpy/reference/ufuncs.html#available-ufuncs) for a complete list of ufuncs available in NumPy.

The ufuncs for basic arithmetic operations are add, subtract, divide, multiply, power, and exp (exponential). However, NumPy uses operator overloading so that we can use mathematical operators (+, -, /, *, and **) directly:

```python
lst = [[1, 2, 3],
        [4, 5, 6]]
ary = np.array(lst)
np.add(ary, 1)
```

```python
ary + 1
```

In addition, NumPy implements several useful unary ufuncs, such as log (natural logarithm), log10 (base-10 logarithm), and sqrt (square root):

```python
np.log(ary)
```

Often, we want to compute the sum or product of array element along a given axis. For this purpose, we can use a ufunc’s reduce operation. By default, reduce applies an operation along the first axis (axis=0):

```python
np.add.reduce(ary)
```

```python
np.add.reduce(ary, axis=1)
```

NumPy also provides specific functions such as sum and product:

```python
ary.sum(axis=0)
```

Note: Both product and sum function compute the product and sum of the entire array if we do not define an axis:

```python
ary.sum()
```

Other useful unary ufuncs are:

* np.mean (computes arithmetic mean or average)
* np.std (computes the standard deviation)
* np.var (computes variance)
* np.sort (sorts an array)
* np.argsort (returns indices that would sort an array)
* np.min (returns the minimum value of an array)
* np.max (returns the maximum value of an array)
* np.argmin (returns the index of the minimum value)
* np.argmax (returns the index of the maximum value)
* np.array_equal (checks if two arrays have the same shape and elements)
